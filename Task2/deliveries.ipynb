```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
deliveries_df = pd.read_csv("deliveries.csv")
```

```python
deliveries_df.groupby('batting_team')['total_runs'].sum().sort_values(ascending=False)
```

```python
deliveries_df.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
deliveries_df.groupby('bowler')['ball'].count().sort_values(ascending=False).head(10)
```

```python
deliveries_df[deliveries_df['batsman_runs'] == 4].groupby('batsman')['batsman_runs'].count().idxmax()
```

```python
deliveries_df[deliveries_df['batsman_runs'] == 6].groupby('batsman')['batsman_runs'].count().idxmax()
```

```python
wide = deliveries_df['wide_runs'].sum()
print ("Wides", wide)
noball = deliveries_df['noball_runs'].sum()
print ("No Balls", noball)
bye = deliveries_df['bye_runs'].sum()
print ("Byes", bye)
legbye = deliveries_df['legbye_runs'].sum()
print ("Leg Byes", legbye)
```

```python
match_id = int(input("Enter match id: "))
total_runs_match = deliveries_df[deliveries_df['match_id'] == match_id]['total_runs'].sum()
print(f"Total runs scored in match {match_id}: {total_runs_match}")
```

```python
avg_runs_per_ball = np.mean(deliveries_df['total_runs'].values)
print(f"Average runs per ball: {avg_runs_per_ball}")
```

```python
max_runs_ball = np.max(deliveries_df['total_runs'].values)
print(f"Maximum runs scored in a single ball: {max_runs_ball}")
min_runs_ball = np.min(deliveries_df['total_runs'].values)
print(f"Minimum runs scored in a single ball: {min_runs_ball}")
```

```python
plt.figure(figsize=(10, 5))
top_batsmen = deliveries_df.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
ax = sns.barplot(x=top_batsmen.values, y=top_batsmen.index, palette="magma")
plt.title("Top 10 Batsmen by Total Runs Scored", fontsize=14, fontweight='bold')
plt.xlabel("Total Runs", fontsize=12)
plt.ylabel("Batsman", fontsize=12)
for p in ax.patches:
    ax.annotate(f"{int(p.get_width())}", (p.get_width() - 300, p.get_y() + p.get_height() / 2.),
                ha='center', va='center', color='white', fontweight='bold')
plt.tight_layout()
plt.show()
```
